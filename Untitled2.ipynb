## Import

### Crop Parameters

```python
dat=pd.read_csv("bosco_giardino_data.csv")
```

```python
for index, row in dat.iterrows():
    if not row['nome_comune'] == 'nan':
        species=PlantSpecies(common_name=row['common_name'],latin_name=row['latin_name'])
        try:
            species.save()
        except:
            pass
```

```python
param=Parameter.objects.filter(code="Thigh").first()
for index, row in dat.iterrows():
    if not row['nome_comune'] == 'nan':
        pl=PlantSpecies.objects.filter(common_name=row['common_name']).first()
        try:
            cp=CropParameter(specie=pl,parameter=param,value=row['thigh'],url_ref="https://mater.cc")
            cp.save()
        except:
            pass
```

```python
"""
Import parameters from CSV table
"""

params=pd.read_csv("params.csv")
for index, row in params.iterrows():
    param=Parameter(code=row['code'], name=row['name'], description=row['description'], measure_unit=row['measure_unit_re0'])
    param.save()
```

```python
cols = ['row_btw', 'row_wtn', 'yield_sqm']
parameter_codes = ['distb', 'distw', 'yield']
for i, col in enumerate(cols):
    param=Parameter.objects.filter(code=parameter_codes[i]).first()
    for index, row in dat.iterrows():
        species=PlantSpecies.objects.filter(common_name=row['common_name']).first()
        if not col=='yield_sqm':
            val=row[col]/100
        else:
            val=row[col]
        paramval=CropParameter(parameter=param, specie=species, value=val, url_ref="http://mater.cc")
        paramval.save()
```

```python
"""
This deletes all the parameters
"""
for col in cols:
    CropParameter.objects.filter(parameter__code=col).delete()
```

### Locations

```python
"""
Create new location-garden-areas
"""
Area.objects.all().delete()
Garden.objects.all().delete()
prode=pd.read_csv("bosco_giardino_prode.csv")
prode=prode[['zona', 'riferimento', 'localizzazione proda', 'lunghezza', 'larghezza']]
locations=prode.drop_duplicates('zona')['zona'].tolist()
for loc in locations:
    l=Location(name=loc, latitude=45, longitude=9)
    #l.save()
for n,row in prode.iterrows():
    l=Location.objects.filter(name=row['zona']).first()
    g=Garden.objects.get_or_create(name=row['riferimento'], location=l)
    a=Area(name=row['localizzazione proda'], garden=g[0], length=row['lunghezza'], width=row['larghezza']/100, temp_offset=0)
    a.save()
```

### Descriptions

```python
"""
Import TP16 characteristics
"""

caratterien=pd.read_csv("codearchivar_characteristics_tp16.csv")
riso=PlantSpecies(common_name="Riso", latin_name="Oryza sativa", plant_type=4)
riso.save()
protocol=Protocol(name="TP16/3", specie=riso,url_ref="http://mater.cc")
protocol.save()
for index, row in caratterien.iterrows():
    t=Trait(numeric_id=row['nchar_id'], description=row['schar_descr'], protocol=protocol)
    t.save()
```

```python
"""
Import TP16 states of expression
"""

classien=pd.read_csv("codearchivar_states_tp16.csv")
for index, row in classien.iterrows():
    trait=protocol.traits.filter(numeric_id=row['nchar_id']).first()
    state=State(numeric_id=row['nclass.id'], description=row['sclass.descr'], trait=trait)
    state.save()
```

```python
archive=pd.read_csv("archive_new.csv")
archive
for i,row in archive.iterrows():
    var=PlantVariety(name=row['svar_name'], species=riso)
    var.save()
    
```

```python
archive.columns
```

```python
for i,row in archive.iterrows():
    var=PlantVariety.objects.filter(name=row['svar_name']).first()
    desc=Description(name="Official",protocol=protocol,variety=var)
    desc.save()
    for n in range(1, 39):
        colname='nchar' + str(n)
        state=desc.available_traits.filter(numeric_id=n).first().state_set.filter(numeric_id=row[colname]).first()
        if state:
            expr=Expression(state_of_expression=state,description=desc)
            expr.save()
```
